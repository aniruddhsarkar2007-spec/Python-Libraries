```python
import time
import numpy as np
size = 8000000

# for list
a=list(range(size))

start_l = time.time()
a = list(map(lambda i : i**2,a))
end_l = time.time() - start_l

#array
a=np.arange(size)
start_a = time.time()
a = [a**2]
end_a = time.time() - start_a

print("Time taken by list:",end_l)
print("Time taken by Array:",end_a)
```
