```python
import time
import numpy as np
size = 8000000

# for list
a=list(range(size))

start_l = time.time()
a = list(map(lambda i : i**2,a))
end_l = time.time() - start_l

#array
a=np.arange(size)
start_a = time.time()
a = [a**2]
end_a = time.time() - start_a

print("Time taken by list:",end_l)
print("Time taken by Array:",end_a)
```

```python
a=np.random.randint(10,20,(3,3))
print(a)
```

```python
# Flatten -- it is used to convert multidimension array into 1 dimension in this there will be no change in the orignal array

flat = a.flatten()
print(flat)
```

```python
# Ravel -- it is used to convert multidimension array into 1 dimension in this the orignal array will also be updated according to the copy one

rav = a.ravel()
rav[2]=100
print(rav)
print(a)
```
